```python
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self._price = 0              #protected attribute
        self.set_price(price)        #validate on creation too

    def get_price(self):
        return self._price

    def set_price(self, new_price):
        if new_price >= 0:
            self._price = new_price
        else:
            print("Price must be 0 or more.")

    def display(self):
        """Display book information."""
        print(f"Title: {self.title}")
        print(f"Author: {self.author}")
        print(f"Price: ${self._price}")


class EBook(Book):
    def __init__(self, title, author, price, size):
        super().__init__(title, author, price)
        self.size = size

    def display(self):
        super().display()
        print(f"File Size: {self.size} MB")


book1 = Book("Powerful Mindset", "Vath CG", 89)
book1.display()

book1.set_price(90)          #valid change
book1.set_price(-10)         #rejected by encapsulation
print("New price:", book1.get_price())

print()
book2 = EBook("Karma", "Kou Sopheap", 78, 314)
book2.display()
```
